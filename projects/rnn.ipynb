```python
#removing the urls
import re
#def rem_url(x):
    ##re.sub(r"http\S+","",x) #pattern ,repl,string
    #return x

#df["review"]=df["review"].apply(rem_url)

def rem_url(x):
    if not isinstance(x, str):  # ← Add this check
        return x
    x = re.sub(r"http\S+", "", x)  # ← Also assign the result back!
    return x


df["review"] = df["review"].apply(rem_url)
```

```python
#removing the punctuation

def punct(x):
    if not isinstance(x,str):
        return x
    x=re.sub(r"[^A-Za-z0-9\s]","",x)#pattern ,repl,string    
    return x

df["review"]=df["review"].apply(punct)
```

```python
def rem_html(x):
    if not isinstance(x,str):
        return x
    x=re.sub(r"<.>","",x)
    return x
df["review"]=df["review"].apply(rem_html)
```

```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
def rem_stopwords(text):
    if not isinstance(text, str):  # ← Add this check
        return text
    tokens=word_tokenize(text)
    stop_words=stopwords.words("english")

    for word in tokens:
        if word in stop_words:
            text.replace(word,"")
        
    return text
df["review"]=df["review"].apply(rem_stopwords)
```
